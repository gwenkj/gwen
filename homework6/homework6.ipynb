```python
mags = np.array([
[2, 3, 5],
[14, 6, 8],
[11, 13, 17],
[7, 10, 13]
])
def prime_rows (mags):
   result = []
   for row in mags:
        has_prime = False
        for mag in row:
            if mag<2:
                continue
            is_prime = True
            for i in range(2,int(mag ** 0.5)+1):
                if mag % i == 0:
                    is_prime = False
                    break
            if is_prime:
                has_prime = True
                break
        if has_prime:
            result.append(row)
   return np.array(result)

print(prime_rows(mags))
```

```python
def checkerboard():
    board = np.array(np.zeros((8,8), dtype=int))
    board[::2,::2] = 1
    board[1::2,1::2] = 1
    return board
print(checkerboard())
```

```python
def reverse_checkerboard():
    board = np.array(np.zeros((8,8), dtype=int))
    board[1::2,::2] = 1
    board[::2,1::2] = 1
    return board
print(reverse_checkerboard())
```

```python
def expansion(text, n):
    result = ""
    for i, character in enumerate(text):
        result += character
        if i<len(text)-1:
            result += " " *n
    return result
expansion("galaxy", 2)
```

```python
def guide_star(stars):
    return np.sort(stars, axis=1)[:,-2]
    
stars = np.array([[1123, 1456, 1789, 1324, 1876],
[1567, 1987, 1678, 1405, 1589],
[1345, 1654, 1523, 1109, 1923],
[1298, 1890, 1367, 1784, 1432],
[1823, 1756, 1489, 1672, 1550]])

print(guide_star(stars))
```
