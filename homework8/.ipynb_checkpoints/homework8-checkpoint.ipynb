```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    return A * np.sin(w * x)
```

```python
x = np.linspace(0, 2*np.pi, 1000)
A = np.array([0.5, 1, 1.5, 2, 2.5])
w = np.array([1, 2, 3, 4, 5])
```

```python
for A, w in zip(A, w):
    print(A, w)
```

```python
fig, axes = plt.subplots(1,3, figsize=(25,5))

axes[0].plot(x, make_sine_wave(x,1,1), label = "data")
axes[0].set_title("X Varied Sine Wave")
axes[0].set_xlabel("x")
axes[0].set_ylabel("Asin(wx)")
axes[0].legend()

axes[1].plot(A, make_sine_wave(1,A,1), label = "data")
axes[1].set_title("A Varied Sine Wave")
axes[1].set_xlabel("A")
axes[1].set_ylabel("Asin(wx)")
axes[1].legend()

axes[2].plot(w, make_sine_wave(1,1,w), label = "data")
axes[2].set_title("W Varied Sine Wave")
axes[2].set_xlabel("W")
axes[2].set_ylabel("Asin(wx)")
axes[2].legend()

plt.show()
```
