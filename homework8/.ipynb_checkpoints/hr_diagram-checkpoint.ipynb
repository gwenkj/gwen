```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt",
skiprows=1)
```

```python
data.shape
# 2642 rows
```

```python
magnitude = data[:,0]
temperature = data[:,1]
```

```python
print(min(magnitude), np.argmin(magnitude))
# 2.24, star 1988 is the brightest
print(max(magnitude), np.argmax(magnitude))
# 12.49, star 264 is the dimmest
```

```python
print(min(temperature), np.argmin(temperature))
# 3653.21, star 1923
print(max(temperature), np.argmax(temperature))
# 10112.78, star 2613
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(log_temperature, log_magnitude, s=50)
plt.scatter(log_temperature[1988], log_magnitude[1988], color="black", s=300, marker="*", label = "brightest star")
plt.scatter(log_temperature[264], log_magnitude[264], color="purple", s=300, marker="*", label = "dimmest star")
plt.scatter(log_temperature[2613], log_magnitude[2613], color="green", s=300, marker="*", label = "hottest star")
plt.scatter(log_temperature[1923], log_magnitude[1923], color="orange", s=300, marker="*", label = "coldest star")
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
plt.legend()
plt.show()
```
